```python
import numpy as np
import pandas as pd

df_clientes = pd.read_csv("users_data.csv")
df_cartoes = pd.read_csv("cards_data.csv")

df_clientes
```

```python
df = pd.merge(df_cartoes, df_clientes, how = "right", left_on = "client_id", right_on = "id")
pd.merge(df_cartoes, df_clientes, how = "right", left_on = "client_id", right_on = "id")

```

```python
#escolhi a media de idade por cartoes sem repetir ids das mesmas pessoas, 1 pessoa so pode aparecer apenas uma vez, o cartao escolhido é o mastercard
df_mastercard = df[df["card_brand"] == "Mastercard"]

# Contar quantos IDs estão repetidos(mastercard)
total_duplicados = df_mastercard.duplicated(subset=["client_id"]).sum()
print(f"Total de ids duplicados: {total_duplicados}")

df_mastercard_unicos = df_mastercard.drop_duplicates(subset=["client_id"])

media_idade = df_mastercard_unicos["current_age"].mean()
print(f"Média de idade (Mastercard): {media_idade:.2f}")
```

```python
grupo_cartoes = df.groupby("card_type")

variabilidade_robusta = grupo_cartoes["current_age"].agg(lambda x: x.quantile(0.75) - x.quantile(0.25))

print("Variabilidade robusta (IQR) da idade por tipo de cartão:")
print(variabilidade_robusta)

#a diferença de idade entre os 25% mais jovens e os 25% mais velhos é de 29 anos
```

```python
tabela_contingencia = pd.pivot_table(df,
                                    values="id_x",
                                    index = "gender",
                                    columns="card_type",
                                    aggfunc = "count")

print("Tabela de Contingência (Ocorrências):")
print(tabela_contingencia)
```

```python
df["yearly_income"] = df["yearly_income"].str.replace("$","").astype(float)

limite_minimo = df.groupby("card_brand")["yearly_income"].min()

print(limite_minimo)

limite_max = limite_minimo.max()

print(limite_max)

renda_media_genero = df.groupby("gender")["yearly_income"].mean()

print(f"Maior valor entre os mínimos das marcas: {limite_max}")
print("\nRenda média por género:")
print(renda_media_genero)
```

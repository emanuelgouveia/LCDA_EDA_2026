```python
import pandas as pd
condutores=pd.read_csv("archive/drivers.csv")
resultado=pd.read_csv("archive/results.csv")

juntar=(pd.merge(condutores, resultado, on= 'driverId'))


subset= (juntar[juntar["driverRef"] =='hamilton'])['position']
subset=pd.to_numeric(subset, errors='coerce') # tornamos position em numero e '\n' fica nan e nao prejudica a media

print(subset.mean()) #não precisa de dropna pq na media, é pre-formatado skipna=True
```

```python

import pandas as pd

resultado = pd.read_csv("archive/results.csv")
construtor = pd.read_csv("archive/constructors.csv")

juntar2 = pd.merge(construtor, resultado, on="constructorId")

juntar2["position"] = pd.to_numeric(juntar2["position"], errors="coerce")
juntar2= juntar2.dropna(subset=["position"])

g = juntar2.groupby("constructorRef")["position"]

iqr = g.quantile(0.75) - g.quantile(0.25) #nao ignora nan

print(iqr)
```

```python
import pandas as pd

corrida = pd.read_csv("archive/races.csv")
circuito = pd.read_csv("archive/circuits.csv")

corridapcircuito= pd.merge(corrida, circuito, on='circuitId')

tabela = corridapcircuito.pivot_table(
    values='raceId',
    index='year',
    columns='country',
    aggfunc='count',
    fill_value=0
)

print(tabela)
```

```python
import pandas as pd

resultados = pd.read_csv("archive/results.csv")
pilotos = pd.read_csv("archive/drivers.csv")

df = pd.merge(resultados, pilotos, on="driverId")

df["points"] = pd.to_numeric(df["points"], errors="coerce")

maxpontosnacionalidade = (
    df.groupby(["nationality", "driverId"])["points"].sum() #soma os pontos de cada piloto por nacionalidade
      .groupby("nationality").max()) #output do pilot com + pontos de cada nacionalidade

print(maxpontosnacionalidade.sort_values(ascending=False).head())
```
